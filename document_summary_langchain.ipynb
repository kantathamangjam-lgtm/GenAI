```python
!pip install pypdf google-genai

from langchain_community.document_loaders import PyPDFLoader

# Upload your PDF in Colab or reference a path
pdf_path = "/content/sample.pdf"  # Replace with your PDF file path
loader = PyPDFLoader(pdf_path)
docs = loader.load()

print(f"Loaded {len(docs)} pages from the PDF.")
```

```python
!pip install langchain-google-genai
#!pip install pypdf google-genai
#!pip install langchain langchain-community

from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.3,
)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Agent 1: Reader / Extraction Agent
extract_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a meticulous research analyst. Extract the core arguments, data points, and key sections from the provided text."),
    ("human", "Text:\n{text}")
])
extract_chain = extract_prompt | llm | StrOutputParser()

# Agent 2: Synthesizer / Executive Summary Agent
summary_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an executive communicator. Synthesize the extracted notes into a clean, well-structured multi-bullet summary with actionable takeaways."),
    ("human", "Extracted Notes:\n{notes}")
])
summary_chain = summary_prompt | llm | StrOutputParser()

# Combine into a multi-step agent flow
full_text = "\n".join([doc.page_content for doc in docs])

# Step 1 execution (Agent 1)
raw_notes = extract_chain.invoke({"text": full_text[:10000]}) # Slice if document is very long

# Step 2 execution (Agent 2)
final_summary = summary_chain.invoke({"notes": raw_notes})

print("=== FINAL MULTI-AGENT SUMMARY ===")
print(final_summary)
```
