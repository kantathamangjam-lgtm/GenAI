```python
#!pip install pypdf genai
!pip install pypdf google-genai

#!pip install -q -U google-genai

import os
from google import genai
from pypdf import PdfReader
from google.colab import userdata

# Initialize the client (ensure GEMINI_API_KEY is set in your environment)
# 1. Safely retrieve the API key from Colab Secrets
GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')

# 2. Initialize the Gemini client
client = genai.Client(api_key=GEMINI_API_KEY)

# Initialize the Gemini client
# It automatically picks up GEMINI_API_KEY from the environment
#client = genai.Client()


def extract_text_from_pdf(pdf_path: str) -> str:
  """Reads a PDF file and extracts all text."""
  reader = PdfReader(pdf_path)
  text = ""
  for page in reader.pages:
    extracted = page.extract_text()
    if extracted:
      text += extracted + "\n"
  return text


def generate_quiz(pdf_text: str) -> str:
  """Sends text to Gemini to generate a 5-question multiple choice quiz."""
  prompt = (
      "Based on the following text from a PDF, generate a 5-question"
      " multiple-choice quiz. Provide 4 options (A, B, C, D) for each question"
      " and include an answer key at the very end.\n\nText:\n"
      f"{pdf_text[:10000]}"  # Truncate to fit context if very large
  )

  response = client.models.generate_content(
      model="gemini-3.6-flash",
      contents=prompt,
  )
  return response.text


# Usage
pdf_file = "/content/sample.pdf"  # Replace with your PDF file path

if os.path.exists(pdf_file):
  print("Reading PDF...")
  content = extract_text_from_pdf(pdf_file)

  print("Generating quiz with Gemini...")
  quiz = generate_quiz(content)

  print("\n--- Generated Quiz ---\n")
  print(quiz)
else:
  print(f"File not found: {pdf_file}")
```
